```python
import random

animales = ['perro', 'gato', 'elefante', 'león', 'tigre', 'jirafa', 'oso']

def elegir_animal_aleatorio(lista):
    return random.choice(lista)

animal_elegido = elegir_animal_aleatorio(animales)
print(f"Animal elegido: {animal_elegido}")
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Supongamos que tenemos un array de lecturas de temperatura (en °C) a lo largo del tiempo
temperaturas = np.array([20, 21, 22, 23, 24, 25, 26])
# Y un array de tiempos correspondientes (por ejemplo, horas)
tiempos = np.arange(len(temperaturas)).reshape(-1, 1)

# Creamos y entrenamos el modelo de regresión lineal
modelo = LinearRegression()
modelo.fit(tiempos, temperaturas)

# Predecimos la temperatura para una hora futura (por ejemplo, la hora 7)
hora_futura = np.array([[7]])
temperatura_predicha = modelo.predict(hora_futura)

print(f"Temperatura predicha para la hora {hora_futura[0][0]}: {temperatura_predicha[0]:.2f} °C")
```

```python
import subprocess

# Escribir el código C++ en un archivo temporal
codigo_cpp = """
#include <iostream>
int main() {
    std::cout << "Hola desde C++!" << std::endl;
    return 0;
}
"""

with open("temp.cpp", "w") as f:
    f.write(codigo_cpp)

# Compilar el archivo C++
subprocess.run(["g++", "temp.cpp", "-o", "temp_exe"])

# Ejecutar el programa compilado
resultado = subprocess.run(["./temp_exe"], capture_output=True, text=True)
print("Salida de C++:", resultado.stdout)
```
